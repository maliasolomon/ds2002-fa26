### How to read an error before you fix it

Python's traceback is read **bottom up**. The last line names the exception and the problem; the line above it points at your code.

```
Traceback (most recent call last):
  File "<stdin>", line 2, in <module>
    print(greeting)
NameError: name 'greeting' is not defined
```

Four exceptions cover most of what you will hit this semester:

| Exception | Usually means |
|---|---|
| `NameError` | The variable was never created, or the cell that created it never ran |
| `TypeError` | You mixed types — text where a number belongs is the classic |
| `KeyError` | That column or dictionary key does not exist. Check spelling and case |
| `FileNotFoundError` | The path is wrong for *this* environment |

### Bug 1 — a name that does not exist

````python
greeting = 'hello'
print(greeting)
````

**What was wrong:** The code attempted to print `greeting`, but `greeting` was never assigned a value/created. Because of this, Python doesn't know what `greeting` is referring to, which is why it results in a `NameError`. To fix this, we define `greeting` before printing.



### Bug 2 — text where a number belongs

This should print `30`. Note that `'10' + 20` and `'10' + '20'` fail in different ways — make sure you know which one you are looking at.

````python
total = 10 + 20
print(total)
````

**What was wrong:** Before, `total` was set to `'10' + 20`, attempting to add a string and an integer together. Python is unable to do arithmetic operations between different types like that, which is why it resulted in a `TypeError`. To fix, we remove the quotes around `10`, making both numbers integers and allowing for addition.

### Bug 3 — off the end of a list

````python
items = ['burger', 'poncho', 'hat']
print(items[-1])
````

**What was wrong:** The `items` list only has indices 0-2 (3 items), so `items[3]` will result in an `IndexError` because index 3 doesn't exist. To fix this to print the last item, we would use `items[-1]`, which always refers to the last item even if someone added to the list later.

### Bug 4 — no error, wrong answer

This is supposed to be the average. It prints a number, and the number is wrong.

````python
nums = [2, 4, 6, 8]
average = sum(nums) / len(nums)
print('average:', average)
````

**What was wrong:** Initially, `average` divided the sum of `nums` by 2, rather than the number of items in the list. This would only give a correct average if there were 2 numbers in the list, so in this case it gives the wrong average because there are 4 numbers in the list. To get the correct average, we would divide `sum(nums)` by `len(nums)` which divides by however many items are in the list, giving a correct average no matter how many items are in the list if the list changes.

### Bug 5 — a table that cannot exist

### Bug 6 — the answer changes when you re-run it

Run the next cell three times and watch the total move. Then fix it using what you built Wednesday: leave the source data alone and derive a new column from it. Your fix must give the same total on every run.

````python
sales = pd.DataFrame({
    'game': ['NC State', 'Louisville'],
    'units': [1450, 1610],
})

work = sales.copy()
work['units'] = sales['units'] * 1.05
print('new total:', round(work['units'].sum(), 2))
````

````python
# Proof — this must print the same number three times
for attempt in range(3):
    work = sales.copy()
    work['units'] = sales['units'] * 1.05
    print(f'attempt {attempt}:', round(work['units'].sum(), 2))
````

**What was wrong:** The line `sales['units'] = sales['units'] * 1.05` overwrote the original `units` column with the new adjusted value. Since the source data  changed, re-running the cell multiplied by 1.05 again on top of the already-adjusted numbers, so the total kept growing with each run instead of staying constant. By making and working with a copy, the `sales` data was left untouched, so every time the calculation is redone it starts from the same raw values and produces the same total.

### Stretch

Write a `check_orders(df)` function that raises a clear error if the frame is missing either expected column, has zero rows, or contains a negative value in `units`. Then call it on `orders`. This is the shape of validation you will want in both projects.

````python
def check_orders(df):
    expected_columns = ["game", "units"]
    missing = [col for col in expected_columns if col not in df.columns]
    if missing:
        raise ValueError(f"Missing expected column(s): {missing}")
    if len(df) == 0:
        raise ValueError("The orders DataFrame has zero rows.")
    if (df["units"] < 0).any():
        raise ValueError("The orders DataFrame contains a negative value in units.")

check_orders(orders)
````

`check_orders(df)` validates that the `orders` data frame is safe to use before running calculations on it. It raises a value error if either of the expected columns (game, units) are missing, if the data frame has zero rows, or if units contains a negative value. This function catches bad data early with a specific message rather than us running into problems later, saving time.
